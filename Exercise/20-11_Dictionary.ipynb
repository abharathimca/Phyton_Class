```python
# Dictionary {} | dict()  
# It stores data as key-value pairs.
# It is Ordered. 
# It is Mutable.
# Does not allow Duplicate Key.
# get(), keys() values() items() update() pop() popitem() clear()  
# setdefault() # - returns the value for the specified key if key is in dictionary, else default_value. 
# If default_value is not provided, it returns None.

# get(key,default_value) - returns the value for the specified key if key is in dictionary, else default_value. 
# If default_value is not provided, it returns None.
student_dictionary.clear
student_dictionary = {"Student_id": 1, "First_Name": "Rathi", "Last_Name": "Ram"}
print("student_dictionary : " , student_dictionary)
print(f"type(student_dictionary) : ", type(student_dictionary))
student_dictionary.setdefault("Fees" , 100000)
print("after setdefault : " , student_dictionary)
student_dictionary.update({"Fees":300000, "Grade" : "X"})
print("after updting fees abd new column grade : " , student_dictionary)
print(f"Value of First_Name direct method : ", student_dictionary["First_Name"])
print(f"Value of Last_Name using get: ", student_dictionary.get("Last_Name"))
student_dictionary["Fees"] = 200000
print("update fees : " , student_dictionary)
student_dictionary.pop("Last_Name")
print("Pop(Remove) Last Name  : " , student_dictionary)
student_dictionary.popitem()
print("PopItem(Remove last item) : " , student_dictionary)
print("student_dictionary keys  :",student_dictionary.keys())
print("student_dictionary Values:",student_dictionary.values())
print("student_dictionary Items :",student_dictionary.items())

Students_dictionary = {
    "Student_id": [1,2,3,4,5],
    "Name": ["Ram", "Ram", "Geethu", "Chandran", "Deeji"]
}
print("Original : ",Students_dictionary)
Students_dictionary.get("Grade","Grade VIII")
print("After get :", Students_dictionary )
print("Multi Values to item :", Students_dictionary.items() )

list_Col_Names = ["Student_id", "Student_Name"]
list_Col_Values = [{1,2,3,4,5}, {"Ram", "Ram", "Geethu", "Chandran", "Deeji"}]
joined_dict = dict(zip(list_Col_Names,list_Col_Values))
print("Joined to list using dictionary  :   ",joined_dict)
```
