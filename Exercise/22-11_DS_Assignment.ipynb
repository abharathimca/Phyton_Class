```python
# QUESTION 1: TUPLE DETAILS

# Create a tuple containing the names of five cities.
# Perform the following operations:
# - Print the tuple.
# - Display the city at index 2.
# - Display the last three cities using slicing.
# - Check how many times a selected city appears using count().
# - Find the index of a selected city using index().

tuple_cities = ("New Delhi","Mumbai", "Kolkatta", "Chennai", "Bengalure")
print(f"Print the tuple : {tuple_cities}")
print(f"Display the city at index 2 : {tuple_cities[2]}")
print(f"Check how many times a selected city('Chennai') appears using count() : {tuple_cities.count("Chennai")}")
print(f"""Find the index of a selected city("Chennai") using index() : {tuple_cities.index("Chennai")}""")
```

```python
# QUESTION 2: TUPLE UPDATE THROUGH LIST CONVERSION

# Create a tuple containing the values 10, 20, 30, 40, and 50.
# Since tuples are immutable:
# - Convert the tuple into a list.
# - Replace 30 with 35.
# - Add 60 to the list.
# - Remove the first element.
# - Convert the updated list back into a tuple.
# - Print the final tuple.

tuple_numbers = (10,20,30,40,50)
print (f"tuple_numbers : {tuple_numbers} {type(tuple_numbers)}")
list_numbers = list(tuple_numbers)
print (f"Convert the tuple into a list : {list_numbers} {type(list_numbers)}")
list_numbers[2] = 35
print(f"Replace 30 with 35 : {list_numbers}")
list_numbers.append(60)
print(f"Add 60 to the list : {list_numbers}")
list_numbers.remove(10)
print (f"Remove the first element : {list_numbers}")
tuple_numbers = tuple(list_numbers)
print(f"Convert the updated list back into a tuple : {tuple_numbers}") 
print(f"Print the final tuple : {type(tuple_numbers)}")
```

```python
# QUESTION 3: COMBINE AND PROCESS TUPLES
# Create two tuples:
# colors1 = ("red", "blue", "green")
# colors2 = ("yellow", "blue", "white")
# Perform the following operations:
# - Combine both tuples into a new tuple.
# - Print the combined tuple.
# - Count how many times "blue" appears.
# - Display the combined tuple using slicing to show alternate elements.

tuple_colors1 = ("red", "blue", "green")
tuple_colors2 = ("yellow", "blue", "white")
combined_tuple = tuple_colors1+ tuple_colors2
print(f"Colors in tuple 1 : {tuple_colors1} {type(tuple_colors1)}")
print(f"Colors in tuple 2 : {tuple_colors2} {type(tuple_colors1)}")
print(f"Combine both tuples into a new tuple and print: {combined_tuple} {type(combined_tuple)}")
print(f"""Count how many times "blue" appears : {combined_tuple.count("blue")}""")
print(f"Display the combined tuple using slicing to show alternate elements : {combined_tuple[::2]}")
```

```python
# QUESTION 4: FIND COMMON AND UNIQUE ELEMENTS
# Create two sets:
# morning_batch = {"Arun", "Bala", "Charan", "Divya"}
# evening_batch = {"Bala", "Divya", "Ezhil", "Farah"}
# Perform the following operations:
# - Display students present in either batch.
# - Display students present in both batches.
# - Display students present only in the morning batch.
# - Display students present only in the evening batch.

morning_batch = {"Arun", "Bala", "Charan", "Divya"}
evening_batch = {"Bala", "Divya", "Ezhil", "Farah"}
print("Either batch:", morning_batch | evening_batch)
print("Both batches:", morning_batch & evening_batch)
print("Morning only:", morning_batch - evening_batch)
print("Evening only:", evening_batch - morning_batch)
print("----------------------------------------------")
print("Either batch:", morning_batch.union(evening_batch))
print("Both batches:", morning_batch.intersection(evening_batch))
print("Morning only:", morning_batch.difference(evening_batch))
print("Evening only:", evening_batch.difference(morning_batch))
```

```python
# QUESTION 5: REMOVE DUPLICATE NAMES
# Create a list containing the following names, including duplicates:
# names = ["Ravi", "Priya", "Ravi", "Kumar", "Priya", "Anu", "Kumar"]
# Convert the list into a set and print the unique names.
# Then convert the set back into a list and print the list.

names_list = ["Ravi", "Priya", "Ravi", "Kumar", "Priya", "Anu", "Kumar"]
print(f"Name list : {names_list} {type(names_list)}")
names_set = set(names_list)
print(f"Converted the list into a set and print the unique names : {names_set} {type(names_set)}")
names_list = list(names_set)
print(f"Then convert the set back into a list and print the list : {names_list} {type(names_list)}")
```

```python
# QUESTION 6: UPDATE PRODUCT SETS
# Create two sets:
# available_products = {"Laptop", "Mouse", "Keyboard"}
# new_products = {"Monitor", "Mouse", "Printer"}
# Perform the following operations:
# - Add "Webcam" to available_products.
# - Remove "Keyboard" from available_products.
# - Combine available_products and new_products using the union operator.
# - Find the products common to both sets using the intersection operator.
# - Find the products available only in available_products using the difference operator.

set_available_products = {"Laptop", "Mouse", "Keyboard"}
set_new_products = {"Monitor", "Mouse", "Printer"}

print(f"available products : {set_available_products} {type(set_available_products)}")
print(f"new products : {set_new_products} {type(set_new_products)}" )
lst_available_products = list(set_available_products)
lst_available_products.append("Webcam")
print(f"""Add "Webcam" to available_products : {lst_available_products} {type(lst_available_products)}""")
lst_available_products.remove("Keyboard")
print(f"""Remove "Keyboard" from available_products : {lst_available_products} {type(lst_available_products)}""")
set_available_products = set(lst_available_products)
print(f"available products : {set_available_products} {type(set_available_products)}")
print(f"new products : {set_new_products} {type(set_new_products)}" )
print(f"Combine available_products and new_products using the union operator : {set_available_products.union(set_new_products)}")
print(f"Find the products common to both sets using the intersection operator : {set_available_products.intersection(set_new_products)}")
print(f"Find the products available only in available_products using the difference operator : {set_available_products.difference(set_new_products)}")
```

```python
# QUESTION 9: MANAGE A SHOPPING CART DICTIONARY
# Create a dictionary for a shopping cart:
# cart = {"Pen": 10, "Notebook": 50, "Bag": 500}
# Perform the following operations:
# - Add "Pencil" with a value of 5.
# - Update the value of "Notebook" to 60 using update().
# - Retrieve the value of "Bag" using get().
# - Remove "Pen" using pop() and display the removed value.
# - Display the remaining dictionary.
# - Use clear() to remove all items and print the empty dictionary.

cart = {"Pen": 10, "Notebook": 50, "Bag": 500}
print(f"Cart : {cart} {type(cart)}")
# cart.update({"Pencil":10})
cart.setdefault("Pencil",5)
print(f"""Add "Pencil" with a value of 5 : {cart} {type(cart)}""")
cart.update({"Notebook":60})
print(f"""Update the value of "Notebook" to 60 using update(): {cart} {type(cart)}""")
print(f"""Retrieve the value of "Bag" using get(): {cart.get("Bag")}""")
print(f"""Remove "Pen" using pop() and display the removed value: {cart.pop("Pen")}""")
print(f"Display the remaining dictionary: {cart} {type(cart)}")
cart.clear()
print(f"Use clear() to remove all items and print the empty dictionary: {cart} {type(cart)}")
```

```python
# QUESTION 10: CREATE A SUBJECT MARKS DICTIONARY USING ZIP
# Create the following two lists:
# subjects = ["Tamil", "English", "Math", "Science"]
# marks = [90, 85, 95, 88]
# Perform the following operations:
# - Create a dictionary using dict(zip(subjects, marks)).
# - Display the dictionary.
# - Use setdefault() to add "Social" with a mark of 80.
# - Use setdefault() on the existing key "Math" with a mark of 100.
# - Print the dictionary and observe whether the existing Math mark changes.

subjects = ["Tamil", "English", "Math", "Science"]
marks = [90, 85, 95, 88]
dict = dict(zip(subjects,marks))
print(f"Display the dictionary: {dict}")
print(f"""Use setdefault() to add "Social" with a mark of 80 : {dict.setdefault("Social",80)}""")
print(f"""Use setdefault() on the existing key "Math" with a mark of 100 : {dict.setdefault("Math",100)}""")
print(f"Print the dictionary and observe whether the existing Math mark changes : {dict}")
print("OBSERVED MATH MARK HAS NOT CHANGED BY SETDEFAULT METHOD")
```
