```python
# QUESTION 1: TUPLE DETAILS

# Create a tuple containing the names of five cities.
# Perform the following operations:
# - Print the tuple.
# - Display the city at index 2.
# - Display the last three cities using slicing.
# - Check how many times a selected city appears using count().
# - Find the index of a selected city using index().

tuple_cities = ("New Delhi","Mumbai", "Kolkatta", "Chennai", "Bengalure")
print(f"Print the tuple : {tuple_cities}")
print(f"Display the city at index 2 : {tuple_cities[2]}")
print(f"Check how many times a selected city('Chennai') appears using count() : {tuple_cities.count("Chennai")}")
print(f"""Find the index of a selected city("Chennai") using index() : {tuple_cities.index("Chennai")}""")
```

```python
# QUESTION 2: TUPLE UPDATE THROUGH LIST CONVERSION

# Create a tuple containing the values 10, 20, 30, 40, and 50.
# Since tuples are immutable:
# - Convert the tuple into a list.
# - Replace 30 with 35.
# - Add 60 to the list.
# - Remove the first element.
# - Convert the updated list back into a tuple.
# - Print the final tuple.

tuple_numbers = (10,20,30,40,50)
print (f"tuple_numbers : {tuple_numbers} {type(tuple_numbers)}")
list_numbers = list(tuple_numbers)
print (f"Convert the tuple into a list : {list_numbers} {type(list_numbers)}")
list_numbers[2] = 35
print(f"Replace 30 with 35 : {list_numbers}")
list_numbers.append(60)
print(f"Add 60 to the list : {list_numbers}")
list_numbers.remove(10)
print (f"Remove the first element : {list_numbers}")
tuple_numbers = tuple(list_numbers)
print(f"Convert the updated list back into a tuple : {tuple_numbers}") 
print(f"Print the final tuple : {type(tuple_numbers)}")
```

```python
# QUESTION 3: COMBINE AND PROCESS TUPLES
# Create two tuples:
# colors1 = ("red", "blue", "green")
# colors2 = ("yellow", "blue", "white")
# Perform the following operations:
# - Combine both tuples into a new tuple.
# - Print the combined tuple.
# - Count how many times "blue" appears.
# - Display the combined tuple using slicing to show alternate elements.

tuple_colors1 = ("red", "blue", "green")
tuple_colors2 = ("yellow", "blue", "white")
combined_tuple = tuple_colors1+ tuple_colors2
print(f"Colors in tuple 1 : {tuple_colors1} {type(tuple_colors1)}")
print(f"Colors in tuple 2 : {tuple_colors2} {type(tuple_colors1)}")
print(f"Combine both tuples into a new tuple and print: {combined_tuple} {type(combined_tuple)}")
print(f"""Count how many times "blue" appears : {combined_tuple.count("blue")}""")
print(f"Display the combined tuple using slicing to show alternate elements : {combined_tuple[::2]}")
```
