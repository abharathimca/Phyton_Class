```python
# This file is created for datatype testing and validation.
# Created by Bharathi on 2026-11-05

first_name = "Bharathi";
last_name = "A";
gender = "Female";
age = 25;
height = 5.6;

def is_male(gender):
    return gender == "Male"

print ("Default print statement for the given data types");
print (first_name, last_name, "is a", gender, "aged", age, "years and has a height of", height, "feet." , "Is Male = ", is_male(gender));

print ("\nF print statement for the given data types");
print (f"{first_name} {last_name} is a {gender} aged {age} years and has a height of {height} feet. Is Male = {is_male(gender)}");

print ("\nF print statement for the identify data types in next line in single line");
print (f"{type(first_name)} {first_name}\n{type(last_name)} {last_name}\n{type(gender)} {gender}\n{type(age)} {age}\n{type(height)} {height}\n{type(is_male(gender))} {is_male(gender)}""");
print ("\nF print statement for the identify data types in next line in multi line");
print (f"""{type(first_name)} {first_name} 
{type(last_name)} {last_name}
{type(gender)} {gender}
{type(age)} {age}
{type(height)} {height}
{type(is_male(gender))} {is_male(gender)}""");

```
