```python
# # AND
# 1 and 1 = true 
# 1 and 0 = False
# 0 and 0 = False
# 0 and 1 = false 

a=5
b=10
c=15
print(a<b and b<c)
print(a<b and b>c)
print(a>b and b<c)
print(a>b and b>c)

# # OR

# 1 or 1 = True
# 1 or 0 = True
# 0 or 1 = true 
# 0 or 0 = False

print(a<b or b<c)
print(a<b or b>c)
print(a>b or b<c)
print(a>b or b>c)
```

```python
# """ arithmetic operator
# Operator	Description		Example
# +		    Addition		10 + 5 = 15
# -		    Subtraction		10 - 5 = 5
# *		    Multiplication		10 * 5 = 50
# /		    Division		10 / 5 = 2.0
# //		    Floor Division		10 // 3 = 3
# %		    Modulus (Remainder)	10 % 3 = 1
# **		power operator		2**3 = 8 """

first_value = input("Enter first number:")
second_value = input("Enter second number:")
print("First number is ",first_value)
print("Second number is ",second_value)

first_value = int(first_value)
second_value = int(second_value)

print("Addition: ",first_value + second_value)
print("Subtraction: ",first_value - second_value)
print("Multiplication: ",first_value * second_value)
print("Division: ",round(float(first_value / second_value), 2))
print("Floor Division: ",first_value // second_value)
print("Modulus: ",first_value % second_value)
print("Power:",first_value ** second_value)
```

```python
# Operator	Example		Same As
# =		x = 5	
# +=		x += 3		x = x + 3
# -=		x -= 3		x = x - 3
# *=		x *= 3		x = x * 3
# /=		x /= 3		x = x / 3
# %=		x %= 3		x = x % 3
# //=		x //= 3		x = x // 3

original_value = 10
print ("Assignment Operator")
print ("value is ",original_value)

value = original_value
value += 5
print("value after += 5 is ",value)
value = original_value
value =+5
print ("value after =+5 is ",value)

value = original_value
value -= 5
print("value after -= 5 is ",value)
value = original_value
value =-5
print ("value after =-5 is ",value)

value = original_value
value *= 5
print("value after *= 5 is ",value)

value = original_value
value /= 5
print("value after /= 5 is ",value)

value = original_value
value %=5
print ("value after %=5 is ",value)

value = original_value
value //=5
print ("value after //=5 is ",value)
```

```python
# Relational Operators
# --------------------

# Operator	Description
# ==		Equal
# !=		Not Equal
# >		Greater than
# <		Less than
# >=		Greater or Equal
# <=		Less or Equal

first_value = input("Enter first number:")
second_value = input("Enter second number:")
print(f"First number is {first_value}")
print(f"Second number is {second_value}")
print(f"first number Equal to second number {first_value} == {second_value} : {first_value == second_value}")
print(f"first number Not Equal to second number {first_value} != {second_value}: {first_value != second_value}")
print(f"first number Greater than second number {first_value} > {second_value}: {first_value > second_value}")
print(f"first number Less than second number {first_value} < {second_value}: {first_value < second_value}")
print(f"first number Greater or Equal to second number {first_value} >= {second_value}: {first_value >= second_value}")
print(f"first number Less or Equal to second number {first_value} <= {second_value}: {first_value <= second_value}")
```

```python
# Logical Operators
# -----------------

# Operator	Description
# and		Both True
# or		Either True
# not		Reverse the result

english = int(input("Enter English marks: "))
tamil = int(input("Enter Tamil marks: "))
maths = int(input("Enter Maths marks: "))

print("English marks is ",english)
print("Tamil marks is ",tamil)  
print("Maths marks is ",maths)

print("All marks is greater than 35 using 'and' operator: ",english > 35 and tamil > 35 and maths > 35)
print("Any marks is less than 35 using 'or' operator: ",english < 35 or tamil < 35 or maths < 35)
print("All marks is greater than 35 using 'not' operator: ",not(english > 35 and tamil > 35 and maths > 35))
```
