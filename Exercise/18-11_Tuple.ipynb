```python
# tuple ()
# A tuple is like List but we can't change it (immutable)
# It is also ordered and allows duplicates.
# It is faster than List.
```

```python
fruits_tuple = ("apple", "banana", "cherry","apple", "banana")
print("Type of tuple:", type(fruits_tuple))
print("fruit Tuple:", fruits_tuple)
print("Length of tuple:", len(fruits_tuple))
print("Count of apple in tuple:", fruits_tuple.count("apple"))
print("Index of banana in tuple:", fruits_tuple.index("banana"))

# Slicing a tuple
sliced_tuple = fruits_tuple[1:4]
print("Sliced fruits tuple:", sliced_tuple)
print("Sliced fruits tuple (from index 2 to end):", fruits_tuple[2:])
print("reversed tuple:", fruits_tuple[::-1])

vegetables_tuple = ("carrot", "broccoli", "spinach")

# Concatenating two tuples
combined_tuple = fruits_tuple + vegetables_tuple
print("Combined fruits and vegetables tuple:", combined_tuple)

# Repeating a tuple
double_combined_tuple = combined_tuple * 2
print("Combined tuple after repetition:", double_combined_tuple)

# tuple to list
list_fruits = list(fruits_tuple)
print(f"Convert tuple to list for fruits : ", list_fruits )
print("pop used for list", list_fruits.pop())
print(f"list after pop : ", list_fruits)
print(type(list_fruits))

# list to tuple
# new_fruit_tuple = (*list_fruits,)
new_fruit_tuple = tuple(list_fruits)
print("Convert list to tuple after pop : ", new_fruit_tuple)
```
