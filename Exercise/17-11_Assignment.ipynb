```python
# QUESTION 4: STRING MANIPULATION

# Write a Python program to: - Store “python programming” in a variable. -
# Convert it into uppercase. - Convert it into title case. - Find the
# length of the string. - Check whether the string starts with “python”.

variable = "python programming"
uppercase_variable = variable.upper()
title_variable = variable.title()
length_variable = len(variable)
starts_with_python = variable.startswith("python")

print(f"Original: {variable}")
print(f"Uppercase: {uppercase_variable}")
print(f"Title Case: {title_variable}")
print(f"Length: {length_variable}")
print(f"Starts with 'python': {starts_with_python}")
print(f"Original: {variable} Uppercase: {uppercase_variable} Title Case: {title_variable} Length: {length_variable} Starts with 'python': {starts_with_python}")
```

```python
# QUESTION 5: REVERSE AND EXTRACT A STRING

# Write a Python program using the following string:
# name = “Python Programming”
# Perform the following operations: - Print the first 6 characters. -
# Print the last 11 characters. - Extract “Programming” using slicing. -
# Reverse the complete string using slicing. - Print every second
# character.
name = "Python Programming"
print(f"First 6 characters: {name[0:6]}")
print(f"Last 11 characters: {name[-11:]}")
print(f"Extracted 'Programming': {name[7:18]}")
print(f"Reversed string: {name[::-1]}")
print(f"Every second character: {name[::2]}")
```

```python
# QUESTION 3: SIMPLE CALCULATOR USING OPERATORS
# Write a Python program to: - Get two numbers from the user. - Perform
# addition, subtraction, multiplication, division, floor division, and
# modulus. - Display all results.

# Example: Enter first number: 20 Enter second number: 6
# Addition: 26 Subtraction: 14 Multiplication: 120 Division: 3.333 Floor
# Division: 3 Modulus: 2

first_number = int(input("Enter first number: "))
second_number = int(input("Enter second number: "))

addition = first_number + second_number
subtraction = first_number - second_number
multiplication = first_number * second_number
division = first_number / second_number
floor_division = first_number // second_number
modulus = first_number % second_number

print(f"Addition: {addition}")
print(f"Subtraction: {subtraction}")
print(f"Multiplication: {multiplication}")
print(f"Division: {float(division):.3f}")
print(f"Floor Division: {floor_division}")
print(f"Modulus: {modulus}")
```

```python
# QUESTION 2: ELECTRICITY BILL CALCULATION
# Write a Python program to calculate an electricity bill based on the
# number of units consumed.
# Conditions: - If units are less than or equal to 100, charge Rs. 2 per
# unit. - If units are greater than 100 and less than or equal to 200,
# charge Rs. 3 per unit. - If units are greater than 200, charge Rs. 5 per
# unit.

# Display the total electricity bill.
# Example: Enter electricity units: 150 Total Electricity Bill: Rs. 450
# Note: Apply the applicable rate to all units consumed, rather than
# slab-wise billing.

units = int(input("Enter electricity units: "))
if units <= 100:
    bill = units * 2
elif units <= 200:
    bill = units * 3    
else:
    bill = units * 5

print(f"Total Electricity Bill: Rs. {bill}")
```

```python
# QUESTION 3: ATM WITHDRAWAL SYSTEM
# Write a Python program to simulate a simple ATM withdrawal.
# Conditions: - Initial balance = Rs. 10,000. - Get the withdrawal amount
# from the user. - Check whether the withdrawal amount is greater than
# zero. - Check whether sufficient balance is available. - If sufficient
# balance is available, deduct the amount and display the remaining
# balance. - Otherwise, display an insufficient balance message.
# Use nested if and else statements.

initial_balance = 10000
withdrawal_amount = int(input("Enter withdrawal amount: "))
if withdrawal_amount > 0:
    if withdrawal_amount <= initial_balance:
        initial_balance -= withdrawal_amount
        print(f"Withdrawal successful. Remaining balance: Rs. {initial_balance}")
    else:
        print("Insufficient balance.")
```

```python
# QUESTION 2: LIST MODIFICATION
# Write a Python program using the following list:
# numbers = [10, 20, 30, 40, 50]
# Perform the following operations: - Change the first element to 100. -
# Insert 25 at index 2. - Add 60 at the end of the list. - Remove the
# value 40. - Extend the list with [70, 80]. - Print the final list.

numbers = [10, 20, 30, 40, 50]
numbers[0] = 100
numbers.insert(2, 25)
numbers.append(60)
numbers.remove(40)
numbers.extend([70, 80])
print("Final list:", numbers)
```

```python
# QUESTION 3: LIST OPERATIONS AND CALCULATIONS
# Write a Python program using the following list:
# marks = [85, 60, 95, 70, 80]
# Perform the following operations: - Sort the list in ascending order. -
# Sort the list in descending order. - Find the maximum mark. - Find the
# minimum mark. - Calculate the total marks. - Calculate the average
# marks. - Check whether all marks are greater than zero using all().

marks = [85, 60, 95, 70, 80]
marks.sort()
print("Ascending order:", marks)
sorted_marks = sorted(marks, reverse=True)
print("Descending order:", sorted_marks)
print("Maximum mark:", max(marks))
print("Minimum mark:", min(marks))
print("Total marks:", sum(marks))
print("Average marks:", sum(marks) / len(marks))
print("All marks are greater than zero:", all(marks))
```
