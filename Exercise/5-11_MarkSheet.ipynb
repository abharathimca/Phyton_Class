```python
import random
import string
from abc import ABC, abstractmethod
line = "*"
first_name = "Rathi"
last_name = "Ram"

total_marks = 0
number_of_subjects = 5
list_of_results = []
for increment in range(80):
    line += "-"

# print(f"""Full Name = {first_name + last_name}""")
print(f"""Full Name = {first_name} {last_name}""")
print(line)
print(f"|\t S.No \t|\t Subject \t|\t Marks \t|\t Result \t|")
print(line)
for subjects in range(number_of_subjects):
    random_name_length = random.randint(8, 10)
    random_letters = "".join(random.choices(string.ascii_letters, k=random_name_length)).upper()
    random_number = random.randint(35, 100)
    total_marks += random_number
    result = 'Pass' if random_number >= 40 else 'Fail'
    list_of_results.append(result)
    print(f"|\t {subjects + 1} \t|\t {random_letters} \t|\t {random_number} \t|\t {result} \t\t|")
    print(line)


percentage = (total_marks / (number_of_subjects * 100)) * 100

def calculate_grade(percentage: float) -> str:
    if percentage >= 84:
        return "Outstanding"
    if percentage >= 70:
        return "A"
    if percentage >= 60:
        return "B"
    if percentage >= 50:
        return "C"
    return "D"

grade = calculate_grade(percentage)

print(f"|\t\t\t Total Marks \t|\t {total_marks}  \t|\t {'Fail' if 'Fail' in list_of_results else 'Pass'} \t\t|")
print(line)

print(f"|\t\t\t Percentage \t|\t {percentage:.2f} \t|\t {grade.lower()} \t\t|")
print(line)
```
